```python
import pandas as pd
import numpy as np
import os
from pandas import DataFrame
from sekwencyjnosc_koncentracji import s_kon
```

```python
data = pd.read_csv('d2-R_data.csv',delimiter='\t')
data
```

```python
rows = []

for _, row in data.iterrows():
    b1,b2,b3,b4 = s_kon(row['wiek'], row['k1'], row['k2'], row['k3'], row['k4'])
    rows.append({'ID': row['ID'], 'P': row['P'],
                 'blok1': b1, 'blok2': b2, 'blok3': b3, 'blok4': b4})

sKon_results = pd.DataFrame(rows)
sKon_results
```
